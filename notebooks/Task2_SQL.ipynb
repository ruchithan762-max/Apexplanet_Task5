```python
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine("sqlite:///../data/superstore.db")

query = """
SELECT *
FROM superstore
LIMIT 5;
"""

result = pd.read_sql(query, engine)

result
```

```python
query = """
SELECT Category,
       SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY Category
ORDER BY Total_Sales DESC;
"""

result = pd.read_sql(query, engine)

result
```

```python
query = """
CREATE TABLE IF NOT EXISTS customers AS
SELECT DISTINCT Customer_ID, Customer_Name
FROM superstore;
"""

with engine.begin() as connection:
    connection.exec_driver_sql(query)

print("Customer table created")
```

```python
query = """
SELECT s.Order_ID,
       s.Customer_ID,
       c.Customer_Name,
       s.Sales
FROM superstore s
JOIN customers c
ON s.Customer_ID = c.Customer_ID
LIMIT 10;
"""

result = pd.read_sql(query, engine)
result
```

```python
query = """
SELECT Product_Name, Sales
FROM superstore
WHERE Sales > (
    SELECT AVG(Sales)
    FROM superstore
)
LIMIT 10;
"""

result = pd.read_sql(query, engine)
result
```

```python
query = """
WITH category_sales AS (
    SELECT Category,
           SUM(Sales) AS Total_Sales
    FROM superstore
    GROUP BY Category
)
SELECT *
FROM category_sales
ORDER BY Total_Sales DESC;
"""

result = pd.read_sql(query, engine)
result
```

```python
query = """
SELECT Product_Name,
       Sales,
       ROW_NUMBER() OVER (ORDER BY Sales DESC) AS Row_Number
FROM superstore
LIMIT 10;
"""

result = pd.read_sql(query, engine)
result
```

```python
query = """
SELECT Product_Name,
       Sales,
       RANK() OVER (ORDER BY Sales DESC) AS Sales_Rank
FROM superstore
LIMIT 10;
"""

result = pd.read_sql(query, engine)
result
```

```python
query = """
SELECT Order_Date,
       Sales,
       LAG(Sales) OVER (ORDER BY Order_Date) AS Previous_Sales
FROM superstore
LIMIT 10;
"""

result = pd.read_sql(query, engine)
result
```

```python
query = """
SELECT Order_Date,
       Sales,
       LEAD(Sales) OVER (ORDER BY Order_Date) AS Next_Sales
FROM superstore
LIMIT 10;
"""

result = pd.read_sql(query, engine)
result
```

```python
query = """
CREATE VIEW IF NOT EXISTS category_sales AS
SELECT Category,
       SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY Category;
"""

with engine.begin() as connection:
    connection.exec_driver_sql(query)

print("View created")
```

```python
query = """
SELECT *
FROM category_sales
ORDER BY Total_Sales DESC;
"""

result = pd.read_sql(query, engine)
result
```

Q1. What are the top 5 products by sales?

Q2. What is the monthly sales trend?

```python
query = """
SELECT strftime('%Y-%m', Order_Date) AS Month,
       SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY Month
ORDER BY Month;
"""

result = pd.read_sql(query, engine)
result
```

Q10. Which orders made a loss?

```python
query = """
SELECT Order_ID,
       Product_Name,
       Sales,
       Profit
FROM superstore
WHERE Profit < 0
ORDER BY Profit ASC
LIMIT 20;
"""

result = pd.read_sql(query, engine)
result
```
