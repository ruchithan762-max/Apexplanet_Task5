```python
missing = df.isnull().sum()

print("Missing values:")
print(missing[missing > 0])
```

```python
# Fill missing categorical values with the most frequent value

df["Customer_Name"] = df["Customer_Name"].fillna(df["Customer_Name"].mode()[0])
df["State"] = df["State"].fillna(df["State"].mode()[0])

print("Missing values after cleaning:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
# Identify numerical columns
numeric_cols = df.select_dtypes(include=np.number).columns

print("Numerical columns:")
print(numeric_cols)

print("\nOutliers using IQR:")

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower) | (df[col] > upper)).sum()

    print(f"{col}: {outliers} outliers")
```

```python
# Handle outliers using IQR capping

for col in ["Sales", "Profit"]:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[col] = df[col].clip(lower=lower, upper=upper)

print("Outliers handled using IQR capping.")
```

```python
print("Dataset shape after cleaning:", df.shape)

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
# Histograms for numerical columns

numeric_cols = ["Quantity", "Discount", "Sales", "Profit"]

df[numeric_cols].hist(
    figsize=(12, 8),
    bins=20
)

plt.suptitle("Distribution of Numerical Variables")
plt.tight_layout()
plt.show()
```

```python
# Boxplots for numerical columns

plt.figure(figsize=(12, 6))

sns.boxplot(data=df[["Quantity", "Discount", "Sales", "Profit"]])

plt.title("Boxplots of Numerical Variables")
plt.xlabel("Variables")
plt.ylabel("Values")
plt.show()
```

```python
# Sales by Category

category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
sns.barplot(x=category_sales.index, y=category_sales.values)

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Correlation Heatmap

correlation = df[["Quantity", "Discount", "Sales", "Profit"]].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
# Monthly Sales Trend

monthly_sales = (
    df.groupby(df["Order_Date"].dt.to_period("M"))["Sales"]
    .sum()
)

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Key EDA findings

print("1. Sales by Category:")
print(df.groupby("Category")["Sales"].sum().sort_values(ascending=False))

print("\n2. Profit by Category:")
print(df.groupby("Category")["Profit"].sum().sort_values(ascending=False))

print("\n3. Sales by Region:")
print(df.groupby("Region")["Sales"].sum().sort_values(ascending=False))

print("\n4. Most Profitable Sub-Categories:")
print(df.groupby("Sub_Category")["Profit"].sum().sort_values(ascending=False).head(5))

print("\n5. Top 5 Products by Sales:")
print(df.groupby("Product_Name")["Sales"].sum().sort_values(ascending=False).head(5))
```
